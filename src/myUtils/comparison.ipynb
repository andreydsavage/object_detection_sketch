```python
import pandas as pd
import json
import glob
import re

# Загрузка всех файлов
files = glob.glob('../../data/*/*_*_stats.json')
files
data = []

for f in files:
    with open(f, 'r') as file:
        stats = json.load(file)
        stats['file'] = f
        data.append(stats)

# Создание DataFrame
df = pd.DataFrame(data)

def extract_model_name(filepath):
    """
    Извлекает имя модели из пути к файлу
    Пример: '../../data/output/yolo_cuda_stats.json' -> 'yolo'
            '../../data/output/frcnn_onnx_cuda_stats.json' -> 'frcnn_onnx'
    """
    # Получаем имя файла без пути
    filename = filepath.split('/')[-1]
    # Убираем '_stats.json' и возможные суффиксы устройств
    model_name = filename.replace('_stats.json', '')
    # Убираем суффиксы _cuda, _cpu в конце
    model_name = re.sub(r'_(cuda|cpu)$', '', model_name)
    return model_name

# Применяем функцию
df['model_name'] = df['file'].apply(extract_model_name)

# Результат
print(df[['model_name', 'fps', 'mean']])
```
